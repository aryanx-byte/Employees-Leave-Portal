## Employees Leave Portal - Data Models

This section outlines the data structures for the employees' leave portal, representing the database schema using pandas DataFrames.

```python
import pandas as pd
```

### 1. Employee Details

This table stores comprehensive information about each employee.

```python
employees_details = pd.DataFrame(columns=[
    'employee_id',
    'first_name',
    'last_name',
    'email',
    'department',
    'position',
    'hire_date'
])

display(employees_details.head())
```

### 2. Login Credentials

This table holds the login information for each employee, linking to their `employee_id`.

```python
login_credentials = pd.DataFrame(columns=[
    'employee_id',
    'username',
    'password_hash',
    'role' # e.g., 'employee', 'admin', 'hr'
])

display(login_credentials.head())
```

### 3. Leave Types

This lookup table defines the different types of leave available to employees.

```python
leave_types = pd.DataFrame(columns=[
    'leave_type_id',
    'type_name', # e.g., 'personal', 'emergency', 'medical'
    'description'
])

display(leave_types.head())
```

### 4. Leave Requests

This table records each leave application made by an employee.

```python
leave_requests = pd.DataFrame(columns=[
    'request_id',
    'employee_id',
    'leave_type_id',
    'start_date',
    'end_date',
    'days_requested',
    'reason',
    'status', # e.g., 'pending', 'approved', 'rejected'
    'applied_date',
    'approved_by',
    'approval_date'
])

display(leave_requests.head())
```

### 5. Approval Statuses

This is a lookup table for the various approval statuses a leave request can have.

```python
approval_statuses = pd.DataFrame(columns=[
    'status_id',
    'status_name', # e.g., 'pending', 'approved', 'rejected'
    'description'
])

display(approval_statuses.head())
```

### 6. Leave Balances

This table tracks the annual leave balance for each employee.

```python
leave_balances = pd.DataFrame(columns=[
    'balance_id',
    'employee_id',
    'year',
    'total_annual_leave_days', # e.g., 30 days
    'leave_taken_days',
    'remaining_leave_days'
])

display(leave_balances.head())
```
